```python
import subprocess
from pathlib import Path
```

```python
def mdtohtml(md: str):
    html = md.replace('md', 'html')
    command = ['pandoc', f'{md}', '-o', f'{html}', '--mathjax', '--syntax-highlighting=none']
    result = subprocess.run(command, capture_output=True, text=True)
    if result.returncode == 0:
        print('Conversion successful.')
    else:
        print('Conversion failed, the error message is as follows:')
        print(result.stderr)
```

```python
def completehtml(html, before, after):
    with open(file=f'../html/{html}', mode='r') as mp:
        m = mp.read()
    with open(file=f'../pages/{html}', mode='w') as ch:
        ch.write(before + m + after)
    print('Complete successful.')
```

```python
def copyindex(homepage:str, index:str):
    with open(file=homepage, mode='r') as hp:
        text = hp.read()
    with open(file=index, mode='w') as ind:
        ind.write(text)
    print('homepage.html copy to index.html')
```

```python
mdpath = Path('../md/')
mdnames = []
for md in mdpath.glob("*.md"):
    mdnames.append(str(mdpath / md.name))
for mdname in mdnames:
    mdtohtml(md=mdname)
```

```python
htmlpath = Path('../html/')
htmlnames = []
for htmlname in htmlpath.glob('*.html'):
    htmlnames.append(htmlname.name)
beforepage = '../html/ba/a-before-page.html'
afterpage = '../html/ba/b-after-page.html'
with open(file=beforepage, mode='r') as bp:
    b = bp.read()
with open(file=afterpage, mode='r') as ap:
    a = ap.read()
for htmlname in htmlnames:
    completehtml(html=htmlname, before=b, after=a)
```

```python
copyindex(homepage='../pages/homepage.html', index='../index.html')
```
